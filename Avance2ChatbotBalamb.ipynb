```python
import re
import unicodedata
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import nltk

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import SnowballStemmer
from nltk.chat.util import Chat, reflections

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)

stemmer = SnowballStemmer('spanish')
stops = set(stopwords.words('spanish')) - {'cuando', 'cuanto', 'donde', 'como', 'cual', 'que', 'no'}

def limpiar_texto(texto):
    # quitar acentos y pasar a minusculas
    texto = texto.lower()
    texto = ''.join(c for c in unicodedata.normalize('NFD', texto) if unicodedata.category(c) != 'Mn')
    texto = re.sub(r'[^a-z0-9\s]', '', texto)

    tokens = word_tokenize(texto, language='spanish')
    tokens = [stemmer.stem(t) for t in tokens if t not in stops and len(t) > 1]
    return " ".join(tokens)

# cargar dataset y aplicar preprocesamiento
df = pd.read_csv('dataset_intenciones.csv')
df['limpio'] = df['texto'].apply(limpiar_texto)

X_train, X_test, y_train, y_test = train_test_split(
    df['limpio'], df['intencion'], test_size=0.25, random_state=42, stratify=df['intencion']
)

# vectorizacion tf-idf y entrenamiento con svm
vec = TfidfVectorizer(ngram_range=(1, 2))
X_train_vec = vec.fit_transform(X_train)
X_test_vec = vec.transform(X_test)

clf = SVC(kernel='linear', probability=True, random_state=42)
clf.fit(X_train_vec, y_train)

# evaluacion del modelo
preds = clf.predict(X_test_vec)
print(f"\nAccuracy: {accuracy_score(y_test, preds):.4f}\n")
print(classification_report(y_test, preds))

# guardar matriz de confusion
clases = sorted(df['intencion'].unique())
cm = confusion_matrix(y_test, preds, labels=clases)

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=clases, yticklabels=clases)
plt.title('Matriz de Confusión - Intenciones Balamb (SVM)')
plt.xlabel('Predicción')
plt.ylabel('Real')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('matriz_confusion_nlp.png', dpi=300)

# exportar modelo para usarlo luego con el modulo de imagenes
joblib.dump(clf, 'modelo_intenciones_svm.pkl')
joblib.dump(vec, 'vectorizador_tfidf.pkl')

# respuestas del chatbot
respuestas = {
    "saludo": "¡Hola! Soy el asistente virtual de la preparatoria Balamb. ¿En qué puedo ayudarte hoy?",
    "inscripcion_costos": "Requisitos de inscripción: certificado de secundaria, acta de nacimiento y comprobante de domicilio. Costos vigentes: Inscripción semestral de $3,500 MXN y colegiatura mensual de $2,800 MXN (pagos del 1 al 10 de cada mes). Trámites en: admisiones.balamb.edu",
    "calendario_escolar": "El inicio del semestre es el 10 de agosto. Los exámenes parciales son del 6 al 11 de octubre, y los finales del 1 al 6 de diciembre (fin de semestre). El calendario completo está en la página oficial.",
    "materiales_libros": "Los materiales y libros de cada semestre se encuentran en libreria.balamb.edu. Se pueden comprar en la librería de la escuela o en línea.",
    "recuperar_contrasena": "Por seguridad y protección de datos personales, para recuperar tu contraseña entra a alumnos.balamb.edu, da clic en 'Recuperar contraseña' y valida tu identidad con tu matrícula y el código enviado a tu correo alterno o teléfono registrado.",
    "envio_tareas": "Si no puedes subir tu tarea a la plataforma, asegúrate de que esté en formato PDF o DOCX y pese menos de 20MB.",
    "soporte_plataforma": "Para problemas en clases en línea, verifica haber entrado con tu correo @balamb.edu, dale permisos de cámara y micrófono al programa o intenta desde otro dispositivo. Las clases grabadas están en el calendario de Canvas.",
    "contacto_soporte": "Para contactar al personal de la escuela, llama o manda WhatsApp a Dirección: 871-721-8787. Horario de atención: 8:00am a 4:00pm de lunes a viernes.",
    "despedida": "¡Fue un placer! Si necesitas más ayuda, no dudes en contactarme.",
    "fallback": "Disculpa, no entendí bien tu consulta. Puedo ayudarte con: inscripciones y costos, calendario escolar, materiales, recuperación de contraseña, entrega de tareas, soporte de Canvas o contacto de oficinas."
}

def clasificar_mensaje(msg, umbral=0.30):
    msg_limpio = limpiar_texto(msg)
    x = vec.transform([msg_limpio])
    probs = clf.predict_proba(x)[0]
    if np.max(probs) < umbral:
        return "fallback", np.max(probs)
    return clf.classes_[np.argmax(probs)], np.max(probs)

# comparacion rapida contra las reglas del avance 1
pares_viejos = [
    [r"(?i).*(buenas tardes|buenas noches|buenos d[ií]as|hola|oye|qu[eé] tal).*", ["Saludo"]],
    [r"(?i).*(inscripci[oó]n|colegiatura|requisitos|reinscripci[oó]n|pagos|costo).*", ["Inscripción"]],
    [r"(?i).*(calendario|ex[aá]menes|vacaciones|fechas|inicio de clases).*", ["Calendario"]],
    [r"(?i).*(libros|lista de materiales|[uú]tiles escolares|materiales|librer[ií]a).*", ["Materiales"]],
    [r"(?i).*(contrase[nñ]a|correo|olvid[eé]|acceso).*", ["Contraseña"]],
    [r"(?i).*(tarea|error al entregar|pdf|plataforma|subir archivo).*", ["Tareas"]],
    [r"(?i).*(canvas|sesi[oó]n|c[aá]mara|en l[ií]nea|micr[oó]fono).*", ["Plataforma"]],
    [r"(?i).*(oficinas|tel[eé]fono|persona|asesor|asesora|director|coordinador|ventanilla).*", ["Contacto"]],
    [r"(?i).*(gracias|nos vemos|hasta luego|bye|adi[oó]s|ya ser[ií]a todo).*", ["Despedida"]],
    [r"(.*)", ["Fallback"]]
]
bot_v1 = Chat(pares_viejos, reflections)

pruebas = [
    "cuanto sale la mensualidad de la prepa",
    "se me fue mi pasword del sistema",
    "no carga mi docuemnto en el portal",
    "a que hora abren la atencion administrativa",
    "cuando son las evaluaciones del semestre"
]

print("\n--- Comparativa Avance 1 vs Avance 2 ---")
for p in pruebas:
    r1 = bot_v1.respond(p)
    r2, conf = clasificar_mensaje(p)
    print(f"Input: '{p}' -> V1: {r1} | V2: {r2} ({conf:.2f})")
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.applications.resnet50 import preprocess_input
from tensorflow.keras import layers, models
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# ruta donde kagglehub descargo las 3 carpetas
data_dir = "/root/.cache/kagglehub/datasets/ritvik1909/document-classification-dataset/versions/1"

img_size = (224, 224)
batch_size = 16

# cargar dataset dividido en entrenamiento (80%) y validacion (20%)
train_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="training",
    seed=42,
    image_size=img_size,
    batch_size=batch_size
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="validation",
    seed=42,
    image_size=img_size,
    batch_size=batch_size
)

clases_visuales = train_ds.class_names
print("Clases detectadas:", clases_visuales)

# construir modelo con transferencia de aprendizaje (ResNet50)
base_model = ResNet50(weights="imagenet", include_top=False, input_shape=(224, 224, 3))
base_model.trainable = False  # congelar pesos preentrenados

inputs = layers.Input(shape=(224, 224, 3))
x = preprocess_input(inputs)
x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)
outputs = layers.Dense(len(clases_visuales), activation="softmax")(x)

modelo_visual = models.Model(inputs, outputs)
modelo_visual.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# entrenar el modelo
historial = modelo_visual.fit(train_ds, validation_data=val_ds, epochs=6)

# evaluar modelo sobre el conjunto de validacion/prueba
y_true = []
y_pred = []

for imagenes, etiquetas in val_ds:
    preds = modelo_visual.predict(imagenes, verbose=0)
    y_true.extend(etiquetas.numpy())
    y_pred.extend(np.argmax(preds, axis=1))

print(f"\nAccuracy Visual: {accuracy_score(y_true, y_pred):.4f}\n")
print(classification_report(y_true, y_pred, target_names=clases_visuales))

# guardar matriz de confusion visual
cm_vis = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(7, 6))
sns.heatmap(cm_vis, annot=True, fmt="d", cmap="Greens", xticklabels=clases_visuales, yticklabels=clases_visuales)
plt.title("Matriz de Confusión - Clasificador de Documentos (ResNet50)")
plt.xlabel("Predicción")
plt.ylabel("Real")
plt.tight_layout()
plt.savefig("matriz_confusion_visual.png", dpi=300)

# guardar modelo entrenado
modelo_visual.save("modelo_visual_resnet50.keras")
```

```python
import glob
import joblib
import numpy as np
import tensorflow as tf
from tensorflow.keras.utils import load_img, img_to_array

# cargar modelos guardados
clf_texto = joblib.load("modelo_intenciones_svm.pkl")
vec_texto = joblib.load("vectorizador_tfidf.pkl")
cnn_visual = tf.keras.models.load_model("modelo_visual_resnet50.keras")

clases_docs = ["email", "resume", "scientific_publication"]

respuestas_documentos = {
    "email": (
        "Documento visual detectado: Correo / Solicitud Administrativa. "
        "Lo he canalizado al área de Atención y Dirección (ventanilla virtual) para dar seguimiento a tu trámite."
    ),
    "resume": (
        "Documento visual detectado: Currículum / Historial Académico. "
        "Este formato corresponde al registro de aspirantes, servicio social o bolsa de trabajo en admisiones.balamb.edu."
    ),
    "scientific_publication": (
        "Documento visual detectado: Artículo / Reporte Académico. "
        "Si deseas entregarlo como tarea en Canvas, recuerda que debe estar en formato PDF o DOCX y pesar menos de 20 MB."
    )
}

def analizar_imagen(ruta_img):
    img = load_img(ruta_img, target_size=(224, 224))
    arr = img_to_array(img)
    arr = np.expand_dims(arr, axis=0)
    probs = cnn_visual.predict(arr, verbose=0)[0]
    idx = np.argmax(probs)
    return clases_docs[idx], probs[idx]

def asistente_multimodal(texto_usuario="", ruta_imagen=None):
    salida = []

    # 1. Procesar componente textual (si hay texto)
    if texto_usuario.strip():
        intencion, conf_txt = clasificar_mensaje(texto_usuario)
        salida.append(f"[Texto -> Intención: {intencion} | Confianza: {conf_txt:.2f}]")
        salida.append(respuestas[intencion])

    # 2. Procesar componente visual (si se adjunta imagen)
    if ruta_imagen:
        tipo_doc, conf_vis = analizar_imagen(ruta_imagen)
        salida.append(f"[Imagen -> Clase: {tipo_doc} | Confianza: {conf_vis:.2f}]")
        salida.append(respuestas_documentos[tipo_doc])

    return "\n".join(salida)

# --- PRUEBAS DE INTEGRACIÓN MULTIMODAL ---
ej_resume = glob.glob("/root/.cache/kagglehub/datasets/ritvik1909/document-classification-dataset/versions/1/resume/*")[0]
ej_pub = glob.glob("/root/.cache/kagglehub/datasets/ritvik1909/document-classification-dataset/versions/1/scientific_publication/*")[0]

print("=== PRUEBA 1: Solo texto (con error ortográfico) ===")
print(asistente_multimodal("cuanto cuesta la colegiatura y que requisitos piden"))

print("\n=== PRUEBA 2: Solo imagen (Currículum / Historial) ===")
print(asistente_multimodal(ruta_imagen=ej_resume))

print("\n=== PRUEBA 3: Multimodal (Texto + Documento académico) ===")
print(asistente_multimodal("hola tengo problemas para subir mi tarea en pdf", ruta_imagen=ej_pub))
```
